## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*
A fixed rule (like "flag any page with impressions>500 and age>180 days") canot weigh multiple signals together or sdjust for interactions between them. The starte pipeline already showed this directly: the hand-written baseline rule hit Precision@50 of 0.240, while a random forest model reached 0.740, nearly 3x better by learning patterns across many signals at once instead of one fixed threshold

```python
print("Baseline rile Precision@50: 0.240")
print("Random forest Precision@50: 0.740")
```
